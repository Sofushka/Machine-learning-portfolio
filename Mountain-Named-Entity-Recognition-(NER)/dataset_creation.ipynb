```python
import pandas as pd 
import json
from datasets import Dataset
```

```python
def dataset_creation(input_dir, output_dir):
    # Load raw JSON dataset generated by ChatGPT
    with open(input_dir, 'r', encoding='utf-8') as f:
        raw_data = json.load(f)

    # Define label mappings
    label2id = {"O": 0, "B-MOUNTAIN": 1, "I-MOUNTAIN": 2}
    id2label = {0: "O", 1:"B-MOUNTAIN", 2:"I-MOUNTAIN"}

    # Convert string tags to numerical label IDs
    formated_data = []
    for item in raw_data:
        formated_data.append({
            "tokens": item['tokens'],
            "ner_tags": [label2id[tag] for tag in item["tags"]]
        })

    # Convert to Hugging Face Dataset format and save
    hf_dataset = Dataset.from_list(formated_data)
    hf_dataset.save_to_disk(output_dir)
```

```python
dataset_creation("data/dataset.json", "data/processed_dataset")
```

```python
dataset_creation("data/test.json", "data/processed_test_dataset")
```
